```python
words=open("names.txt","r").read().splitlines()
```

```python
import torch
import random
```

```python
itos={i+1:s for i,s in enumerate(sorted(set(''.join(words))))}
itos[0]='.'
stoi={s:i for i,s in itos.items()}
```

```python
x1=int(0.8*len(words))
x2=int(0.9*len(words))
train=words[:x1]
dev=words[x1:x2]
test=words[x2:]
```

```python
bigram=torch.zeros((27,27),dtype=torch.int32)
```

```python
bigram.dtype
```

```python
for w in train:
    ch=['.']+list(w)+['.']
    for ch1,ch2 in zip(ch,ch[1:]):
        bigram[stoi[ch1],stoi[ch2]]+=1             
```

```python
P=(bigram+1).float()
```

```python
P/=P.sum(1,keepdim=True)
```

```python
def evaluate_bigram(datasets):
    likelihood=0.0
    n=0
    for w in datasets:
        ch=['.']+list(w)+['.']
        for ch1,ch2 in zip(ch,ch[1:]):
            probs=P[stoi[ch1],stoi[ch2]]
            logits=torch.log(probs)
            likelihood+=logits
            n+=1
    nll = -likelihood
    loss = nll / n
    return (f"loss :{loss:.4f}")
```

```python
print(f" loss of train in bigram model :{evaluate_bigram(train)}")
print(f" loss of dev in bigram model :{evaluate_bigram(dev)}")
print(f" loss of test in bigram model :{evaluate_bigram(test)}")
```

```python
trigram=torch.zeros((27,27,27),dtype=torch.int32)
```

```python
for w in train:
    ch=['.','.']+list(w)+['.']
    for ch1,ch2,ch3 in zip(ch,ch[1:],ch[2:]):
        trigram[stoi[ch1],stoi[ch2],stoi[ch3]]+=1
```

```python
P=(trigram+1).float()
```

```python
P/=P.sum(2,keepdim=True)
```

```python
def evaluate_trigram(datasets):
    likelihood=0.0
    n=0
    for w in datasets:
        ch=['.','.']+list(w)+['.']
        for ch1,ch2,ch3 in zip(ch,ch[1:],ch[2:]):
            probs=P[stoi[ch1],stoi[ch2],stoi[ch3]]
            logits=torch.log(probs)
            likelihood+=logits
            n+=1
    nll = -likelihood
    loss = nll / n
    return (f"loss :{loss:.4f}")
```

```python
print(f" loss of train in triigram model :{evaluate_trigram(train)}")
print(f" loss of dev in triigram model :{evaluate_trigram(dev)}")
print(f" loss of test in triigram model :{evaluate_trigram(test)}")
```
